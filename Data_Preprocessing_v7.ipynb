#Data Preprocessing in Machine learning
O **pré-processamento de dados** é o processo de preparação dos dados brutos, tornando-os adequados para serem utilizados por um modelo de **Machine Learning (Aprendizado de Máquina)**. Ele constitui a **primeira e uma das etapas mais importantes** na construção de um modelo de aprendizado de máquina.

Ao desenvolver um projeto de Machine Learning, nem sempre encontramos dados **limpos, organizados e formatados** de maneira adequada. Na prática, os dados podem apresentar valores ausentes, informações inconsistentes, ruídos, duplicidades ou diferentes formatos.

Antes de realizar qualquer operação ou aplicar um modelo aos dados, é fundamental **limpá-los, organizá-los e estruturá-los adequadamente**. Para isso, utilizamos as técnicas e procedimentos de **pré-processamento de dados**.

Em outras palavras, o pré-processamento tem como objetivo **transformar os dados brutos em um formato adequado para que os algoritmos de Machine Learning possam processá-los de maneira eficiente e produzir resultados mais confiáveis**.


### Padronização (*Standardization*)

A **padronização (*standardization*)** é uma técnica de pré-processamento de dados utilizada para transformar os atributos de um conjunto de dados de modo que apresentem uma **distribuição normal (distribuição Gaussiana)** padronizada.

O objetivo é transformar os dados para que apresentem:

* **média ($μ$) = 0**
* **desvio-padrão ($σ$) = 1**

A transformação é realizada por meio do **escore-z (*z-score*)**:

$$
z = \frac{x-\mu}{\sigma}
$$

onde:

* $(x)$ = valor original;
* $(\mu)$ = média dos dados;
* $(\sigma)$ = desvio-padrão;
* $(z)$ = valor padronizado.

A padronização é particularmente útil em algoritmos **Machine Learning** nos quais a escala das variáveis pode influenciar o processo de aprendizagem. Exemplos incluem **Regressão Linear**, **Regressão Logística**, **SVM**, **K-Means** e redes neurais.

É importante observar que a padronização (*Standardization*) **não transforma** necessariamente os dados originais (o formato do histograma ou a assimetria) em uma **distribuição Gaussiana**. Ela apenas **centraliza os dados em torno de zero e ajusta sua escala para que o desvio-padrão seja igual a 1**. Se os seus dados originais forem totalmente assimétricos ou bimodais, eles continuarão assimétricos ou bimodais após a padronização.

Abaixo o gráfico da distribuição Normal Padrão (também conhecida como Curva de Gauss ou Curva do Sino). Ele ilustra como os dados se distribuem em torno da média (que neste caso é 0) em termos de desvios padrão.

````python
### Exemplo simples
# Considere os valores:
X = [[10],
     [20],
     [30],
     [40],
     [50]]

from sklearn.preprocessing import StandardScaler
import numpy as np

print('X mean:', np.array(X).mean())
print('X std:', np.array(X).std())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print('X scaled:\n', X_scaled)
print('X scaled mean:', X_scaled.mean())
print('X scaled std:', X_scaled.std())
````

Os valores resultantes terão aproximadamente:

$$
\mu = 0
$$

e

$$
\sigma = 1
$$

Assim, valores originalmente em uma escala de **10 a 50** passam a ser representados em uma escala centrada em **zero**, facilitando o processamento por diversos algoritmos de Machine Learning.

##Normalização de dados

**Normalização (*Normalization*)** é um conceito amplo, e existem diferentes maneiras de normalizar os dados. De modo geral, a normalização refere-se ao processo de **ajustar valores medidos em diferentes escalas para uma escala comum**.

Às vezes, a melhor maneira de compreender o conceito é por meio de um exemplo. Para cada tipo de normalização apresentado a seguir, vamos considerar um modelo que busca compreender a relação entre o **preço de uma casa e seu tamanho**.

### Tipos de Normalização

Vamos analisar alguns dos principais tipos de normalização. É importante observar que esta **não é uma lista completa**, pois existem outras técnicas de normalização.

#### Normalização Min-Max

Na **normalização Min-Max**, podemos reescalar os tamanhos das casas para que seus valores fiquem dentro de um intervalo entre **0 e 1**.

Isso significa que:

* o tamanho da menor casa será representado pelo valor **0**;
* o tamanho da maior casa será representado pelo valor **1**;
* os demais tamanhos ficarão entre 0 e 1.

A transformação é realizada por meio da fórmula:

$$
x'=\frac{x-x_{\min}}{x_{\max}-x_{\min}}
$$

---

#### Normalização Logarítmica

A **normalização logarítmica (Log Normalization)** é outra técnica de normalização.

Nesse caso, aplicamos uma **transformação logarítmica** aos preços das casas. Essa técnica ajuda a **reduzir o impacto de valores muito elevados**, especialmente quando existem diferenças significativas entre os preços.

Por exemplo, se algumas casas apresentam preços muito maiores que as demais, a transformação logarítmica pode reduzir a influência desses valores extremos sobre o modelo.

---

#### Escalonamento Decimal

O **escalonamento decimal (Decimal Scaling)** é outra técnica de normalização.

Nesse exemplo, podemos ajustar os tamanhos das casas **deslocando a posição da vírgula decimal**, de modo a tornar os valores menores.

Assim, os tamanhos das casas são transformados para uma escala mais fácil de manipular, **mantendo as diferenças relativas entre os valores**.

Por exemplo:

$$
250 \rightarrow 0,250
$$

$$
500 \rightarrow 0,500
$$

$$
750 \rightarrow 0,750
$$

Os valores tornam-se menores, mas sua relação relativa é preservada.

---

#### Normalização pela Média (Mean Normalization / Mean-Centering)

A **normalização pela média**, também chamada de **centralização pela média (Mean-Centering)**, consiste em ajustar os preços das casas **subtraindo o preço médio de cada valor individual**.

A transformação pode ser representada por:

$$
x'=x-\mu
$$

onde:

* \($x$\) = valor original;
* \($\mu$\) = média dos valores;
* \($x'$\) = valor após a centralização.

Esse processo **centraliza os valores em torno de zero**, permitindo observar quanto cada casa está acima ou abaixo do preço médio.

Por exemplo, se o preço médio das casas for:

$$
\mu = R\$\,500.000
$$

uma casa de:

$$
R\$\,600.000
$$

terá:

$$
600.000-500.000=100.000
$$

enquanto uma casa de:

$$
R\$\,400.000
$$

terá:

$$
400.000-500.000=-100.000
$$

Dessa forma, podemos identificar facilmente quais valores estão **acima ou abaixo da média**, facilitando a interpretação das diferenças relativas entre os preços.


##Normalization vs Standardization
A diferença entre **Normalization** e **Standardization** é uma das mais importantes no **pré-processamento de dados**. Embora as duas técnicas alterem a escala das variáveis, elas fazem isso de maneiras diferentes.

## 1. Normalization — Normalização

A **normalização** geralmente coloca os valores em um **intervalo definido**, sendo muito comum o intervalo:

$$
[0,1]
$$

A técnica mais conhecida é o **Min-Max Scaling**:

$$
\boxed{x'=\frac{x-x_{\min}}{x_{\max}-x_{\min}}}
$$

Assim:

* o menor valor → 0
* o maior valor → 1
* os demais valores ficam entre 0 e 1.

### Exemplo

Considere:

$$
X = [10,\;20,\;30,\;40,\;50]
$$

Temos:

$$
x_{\min}=10
$$

$$
x_{\max}=50
$$

Então:

| Original | Normalizado |
| -------: | ----------: |
|       10 |        0,00 |
|       20 |        0,25 |
|       30 |        0,50 |
|       40 |        0,75 |
|       50 |        1,00 |

No Python:

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_norm = scaler.fit_transform(X)
```

---

# 2. Standardization — Padronização

A **padronização** transforma os dados utilizando a média e o desvio-padrão:

$$
\boxed{z=\frac{x-\mu}{\sigma}}
$$

O resultado possui:

$$
\mu \approx 0
$$

e

$$
\sigma \approx 1
$$

Por exemplo:

| Original | Padronizado |
| -------: | ----------: |
|       10 |       -1,41 |
|       20 |       -0,71 |
|       30 |        0,00 |
|       40 |        0,71 |
|       50 |        1,41 |

No Python:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_std = scaler.fit_transform(X)
```

---

# 3. Comparação direta

| Característica             | **Normalization**                      | **Standardization**               |
| -------------------------- | -------------------------------------- | --------------------------------- |
| Também chamada de          | Min-Max Scaling                        | Z-score Scaling                   |
| Principal objetivo         | Colocar dados em determinado intervalo | Centralizar e ajustar a dispersão |
| Transformação comum        | \([0,1]\)                              | Média 0, desvio-padrão 1          |
| Usa mínimo/máximo?         | ✅ Sim                                  | ❌ Não                             |
| Usa média?                 | ❌ Não                                  | ✅ Sim                             |
| Usa desvio-padrão?         | ❌ Não                                  | ✅ Sim                             |
| Fórmula                    | $\frac{x-x_{min}}{x_{max}-x_{min}}$  | $\frac{x-\mu}{\sigma}$          |
| Limita os valores?         | Sim, normalmente entre 0 e 1           | Não                               |
| Depende de outliers?       | Bastante                               | Também, mas de maneira diferente  |
| Torna os dados Gaussianos? | ❌ Não                                  | ❌ Não                             |
| Média final = 0?           | ❌ Não necessariamente                  | ✅ Sim                             |
| Desvio-padrão final = 1?   | ❌ Não necessariamente                  | ✅ Sim                             |

---

## 4. Uma diferença visual importante

Podemos pensar assim:

```text
NORMALIZATION

Dados originais
     ↓
  [10 ... 50]
     ↓
Transformação Min-Max
     ↓
  [0 ... 1]
```

Enquanto:

```text
STANDARDIZATION

Dados originais
     ↓
[10 ... 50]
     ↓
subtrai a média
     ↓
divide pelo desvio-padrão
     ↓
[-1.41 ... +1.41]
```

A **Normalization** preocupa-se principalmente com o **intervalo/escala**.

A **Standardization** preocupa-se com a **posição dos valores em relação à média e à dispersão**.

---

## 5. Quando utilizar cada uma?

### Normalization

É frequentemente interessante quando o algoritmo trabalha melhor com valores em uma escala limitada.

Exemplos:

* Redes neurais;
* K-Nearest Neighbors (KNN);
* algoritmos baseados em distância;
* algumas aplicações de processamento de imagens.

Por exemplo, uma imagem normalmente possui pixels entre:

$$
0 \leq pixel \leq 255
$$

Podemos transformá-los em:

$$
0 \leq pixel_{norm} \leq 1
$$

usando:

$$
pixel_{norm}=\frac{pixel}{255}
$$

---

### Standardization

É frequentemente utilizada quando queremos remover diferenças de escala entre variáveis e trabalhar com a posição dos valores em relação à média.

É bastante comum em:

* Regressão Linear;
* Regressão Logística;
* SVM;
* PCA;
* K-Means;
* redes neurais;
* métodos baseados em gradiente.

---

## 6. ⚠️ Uma observação importante

Um erro muito comum é pensar:

> **Normalization = transformar em distribuição normal**

Isso está **errado**.

**Normalization** não significa tornar a distribuição dos dados **Normal/Gaussiana**.

Da mesma forma:

> **Standardization = tornar os dados Gaussianos**

também está **errado**.

A Standardization produz:

$$
\mu=0,\quad \sigma=1
$$

mas **não altera necessariamente o formato da distribuição**.

### Resumo para uma aula de Machine Learning

> 🟦 **Normalization:** ajusta os dados para uma escala ou intervalo definido, frequentemente \([0,1]\).

> 🟧 **Standardization:** transforma os dados para que tenham média 0 e desvio-padrão 1.

> 🟩 **Gaussian/Normal Distribution:** descreve o **formato da distribuição** dos dados — é outro conceito.

Essa distinção entre **Normal Distribution × Normalization × Standardization** é especialmente importante para evitar uma confusão terminológica muito comum em cursos introdutórios de Machine Learning.


## Escalonamento Min-Max (Min Max scaling)

Provavelmente, nosso conjunto de dados é composto por atributos que apresentam **escalas diferentes**. Entretanto, não podemos simplesmente fornecer esses dados dessa forma a alguns algoritmos de Machine Learning, sendo necessário realizar um **reescalonamento**.

O **reescalonamento dos dados** garante que os diferentes atributos estejam em uma **escala comparável**. Geralmente, os atributos são reescalonados para o intervalo entre **0 e 1**.

### Definindo um intervalo personalizado

O intervalo de transformação pode ser definido por meio do parâmetro **`feature_range`**.

Por exemplo, para transformar os dados para o intervalo \([-1,1]\):

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(-1, 1))
X_scaled = scaler.fit_transform(X)
```

Assim, o `MinMaxScaler` não está limitado ao intervalo \([0,1]\); podemos especificar o intervalo desejado por meio do parâmetro `feature_range`.


````python
# min max scaling
data_scaler = preprocessing.MinMaxScaler(feature_range=(0, 1))
data_scaled = data_scaler.fit_transform(data2)
print ("Min max scaled data:\n", data_scaled)
````

````python
print ("Min max scaled data Mean =", data_scaled.mean(axis=0))
print ("Min max scaled data Std deviation =", data_scaled.std(axis=0))
````

## L1 Normalization

It may be defined as the normalization technique that modifies the dataset values in a way that **in each row** the sum of the absolute values will always be up to 1. It is also called **Least Absolute Deviations**.

````python
data2
````

Data before L1 normalization:

````python
# sum of each row
data2.sum(axis=1)
````

````python
# L1 normalization
data_normalized = preprocessing.normalize(data2, norm='l1')
print ("L1 normalized data:\n", data_normalized)
# another way
Data_normalizer = preprocessing.Normalizer(norm='l1').fit(data2)
Data_normalized_ = Data_normalizer.transform(data2)
print ("\nL1 normalized data:\n", Data_normalized_)
````

## L2 Normalization
It may be defined as the normalization technique that modifies the
dataset values in a way that in each row **the sum of the squares** will always be up to 1. It is also called **least squares**.

````python
# L2 normalization
data_normalized_l2 = preprocessing.normalize(data2, norm='l2')
print ("L2 normalized data:\n", data_normalized_l2)
````

````python
# square computation
sq=np.square(data_normalized_l2)
sq
````

````python
# least squares
for i in range(5):
  print(sum(sq[i]))
````

````python
# other way to sum the rows
sq.sum(axis=1)
````

## One Hot Encoding
O One-Hot Encoding é uma técnica utilizada para representar variáveis categóricas como vetores binários.

Primeiramente, os valores categóricos são associados a valores inteiros.

Em seguida, cada valor inteiro é representado por um vetor binário, no qual todos os elementos são iguais a 0, exceto a posição correspondente ao valor inteiro, que recebe o valor 1.

````python
# one hot encoding
encoder = preprocessing.OneHotEncoder()
x=np.array([[0], [2], [1], [3], [4]])
print(x)
encoder.fit(x)
encoded_vector = encoder.transform(x).toarray()
print ("\nEncoded vector:\n", encoded_vector)
````

In this example, we will use the encoders from the scikit-learn library. Specifically, the `LabelEncoder` of creating an integer encoding of labels and the `OneHotEncoder` for creating a one hot encoding of integer encoded values.

````python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
# define example
data3 = ['cold', 'cold', 'warm', 'cold', 'hot', 'hot', 'warm', 'cold', 'warm', 'hot']
values = np.array(data3)
print(values)
````

````python
# integer encode
label_encoder = LabelEncoder()
integer_encoded = label_encoder.fit_transform(values)
print(integer_encoded)
````

````python
integer_encoded_reshaped = integer_encoded.reshape(len(integer_encoded), 1)
integer_encoded_reshaped
````

````python
# binary encode
onehot_encoder = OneHotEncoder(sparse_output=False)
onehot_encoded = onehot_encoder.fit_transform(integer_encoded_reshaped)
print(onehot_encoded)
````

````python
# find label of first line
inverted = label_encoder.inverse_transform([np.argmax(onehot_encoded[0, :])])
print(inverted)
````

````python
# Exemplo de pré-processamento
# Contribuição de Marcelo Henrique Jeronymo <marcelo.henrique@uel.br>
#
# Utilize as técnicas de pré-processamento de dados (Normal distribution, Min
# Max scaling, L1 and L2 normalizations) no banco de dados e visualize os resultados após cada etapa
#

import numpy as np
from sklearn import datasets
from sklearn import preprocessing
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs

x_train, y_true, true_centroids = make_blobs(n_samples=400, n_features=2, centers=2, cluster_std=3, random_state=1, return_centers=True)

x_train_norm = preprocessing.scale(x_train)

minmax = preprocessing.MinMaxScaler()
x_train_minmax = minmax.fit_transform(x_train)

l1norm = preprocessing.Normalizer(norm="l1")
x_train_l1norm = l1norm.fit_transform(x_train)

l2norm = preprocessing.Normalizer(norm="l2")
x_train_l2norm = l2norm.fit_transform(x_train)

binary = preprocessing.Binarizer(threshold=1.4)
x_train_bin = binary.fit_transform(x_train)

onehot = preprocessing.OneHotEncoder()
x_train_bin = onehot.fit_transform(x_train)

fig, ax = plt.subplot_mosaic([["unnorm", "unnorm", 0, 1],
                              ["unnorm", "unnorm", 2, 3]], figsize=(12, 6), layout="constrained")


ax["unnorm"].scatter(*x_train.T, c=y_true, cmap="plasma")
ax[0].scatter(*x_train_norm.T, c=y_true, cmap="plasma")
ax[1].scatter(*x_train_minmax.T, c=y_true, cmap="plasma")
ax[2].scatter(*x_train_l1norm.T, c=y_true, cmap="plasma")
ax[3].scatter(*x_train_l2norm.T, c=y_true, cmap="plasma")


titles = ["Unnormalized", "Normalized", "Min Max", "$L_1$ Norm", "$L_2$ Norm"]
for i, axis in enumerate(ax.values()):
    axis.set_title(titles[i])
    axis.grid(lw=0.4, alpha=0.5)
    axis.set_xlabel("$x_1$")
    axis.set_ylabel("$x_2$")

plt.show()
````

````python
# Exemplo de pré-processamento
# Contribuição de Gabriel Delongui Polvani <gabriel.polvani@uel.br>

import numpy as np
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
from sklearn import preprocessing

X, Y = make_blobs(n_samples=400, centers=2, cluster_std=3.00, random_state=1)
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(ncols=2, nrows=2, tight_layout=True, sharex=True, sharey=True)

# Normalização normal-Padrão
X_normal = preprocessing.scale(X)
ax1.scatter(*X_normal[Y==0].T, color='#12629A', alpha=0.6, marker='o')
ax1.scatter(*X_normal[Y==1].T, color='#F247B2', alpha=0.6, marker='x')
ax1.legend(['Classe 1', 'Classe 2'], loc='lower right')

# Normalização MinMax
X_MinMax = preprocessing.MinMaxScaler(feature_range=(0, 1))
X_MinMax = X_MinMax.fit_transform(X)

ax2.scatter(*X_MinMax[Y==0].T, color='#12629A', alpha=0.6, marker='o')
ax2.scatter(*X_MinMax[Y==1].T, color='#F247B2', alpha=0.6, marker='x')
ax2.legend(['Classe 1', 'Classe 2'], loc='lower right')

# Normalização L1
X_L1 = preprocessing.normalize(X, norm='l1')

ax3.scatter(*X_L1[Y==0].T, color='#12629A', alpha=0.6, marker='o')
ax3.scatter(*X_L1[Y==1].T, color='#F247B2', alpha=0.6, marker='x')
ax3.legend(['Classe 1', 'Classe 2'], loc='lower right')

# Normalização L2
X_L2 = preprocessing.normalize(X, norm='l2')

ax4.scatter(*X_L2[Y==0].T, color='#12629A', alpha=0.6, marker='o')
ax4.scatter(*X_L2[Y==1].T, color='#F247B2', alpha=0.6, marker='x')
ax4.legend(['Classe 1', 'Classe 2'], loc='lower right')

ax1.grid(alpha=0.6), ax2.grid(alpha=0.6), ax3.grid(alpha=0.6), ax4.grid(alpha=0.6)

# Adicionando texto para as legendas dos subplots e legendas dos eixos (Coordenadas x e y obtidas empiricamente)
fig.text(0.075, 0.94, 'Standard Normal')
fig.text(0.548, 0.94, 'MinMax (0,1)')
fig.text(0.075, 0.47, 'L1 Normalization')
fig.text(0.548, 0.47, 'Standard Normal')

fig.text(0.51, 0, '$x_1$')
fig.text(0, 0.52, '$x_2$', rotation=90)

plt.show()
````

````python
# Contribuição de João Paulo Monteiro Santana <joao.paulo.santana@uel.br>

# Algoritmo KNN para os seguintes datasets: X, X_normal, X_MinMax, X_L1 e X_L2.
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
from sklearn import preprocessing

X, Y = make_blobs(n_samples=400, centers=2, cluster_std=3.00, random_state=1)

# Subdivisão entre as etapas treinamento e teste
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, Y_train);
Y_pred = knn.predict(X_test)

# Plot para o banco de dados padrão

fig, axs = plt.subplots(ncols=3, nrows=1, figsize=(9,3), tight_layout=True, sharey=True)

axs[0].scatter(X_train[:,0],X_train[:,1],c=Y_train)
axs[0].set_title('Training Set')
axs[0].grid()

axs[1].scatter(X_test[:,0],X_test[:,1],c=Y_test)
axs[1].set_title('Test Set')
axs[1].grid()

axs[2].scatter(X_test[:,0],X_test[:,1],c=Y_pred)
axs[2].set_title('Predict Set')
axs[2].grid()

plt.show()
````

````python
# Agora, testando o algoritmo novamento com os dados pré processados:

# Normalização Normal

X_train_normal, X_test_normal, Y_train_normal, Y_test_normal = train_test_split(X_normal, Y, test_size=0.3, random_state=1)

knn = KNeighborsClassifier()
knn.fit(X_train_normal, Y_train_normal);
Y_pred_normal = knn.predict(X_test_normal)

# Normalização MinMax

X_train_MinMax, X_test_MinMax, Y_train_MinMax, Y_test_MinMax = train_test_split(X_MinMax, Y, test_size=0.3, random_state=1)

knn = KNeighborsClassifier()
knn.fit(X_train_MinMax, Y_train_MinMax);
Y_pred_MinMax = knn.predict(X_test_MinMax)

# Normalização L1

X_train_L1, X_test_L1, Y_train_L1, Y_test_L1 = train_test_split(X_L1, Y, test_size=0.3, random_state=1)

knn = KNeighborsClassifier()
knn.fit(X_train_L1, Y_train_L1);
Y_pred_L1 = knn.predict(X_test_L1)

# Normalização L2

X_train_L2, X_test_L2, Y_train_L2, Y_test_L2 = train_test_split(X_L2, Y, test_size=0.3, random_state=1)

knn = KNeighborsClassifier()
knn.fit(X_train_L2, Y_train_L2);
Y_pred_L2 = knn.predict(X_test_L2)

fig, axs = plt.subplots(ncols=3, nrows=4, figsize=(9, 9), tight_layout=True ,sharey=True, sharex=True)

# Normalização Normal Matriz de Confusão para Métrica

axs[0,0].scatter(X_train_normal[:,0],X_train_normal[:,1],c=Y_train_normal)
axs[0,0].set_title('Training Normal Preprocessing')
axs[0,0].grid()

axs[0,1].scatter(X_test_normal[:,0],X_test_normal[:,1],c=Y_test_normal)
axs[0,1].set_title('Test Normal Preprocessing')
axs[0,1].grid()

axs[0,2].scatter(X_test_normal[:,0],X_test_normal[:,1],c=Y_pred_normal)
axs[0,2].set_title('Predict Normal Preprocessing')
axs[0,2].grid()

# Normalização MinMax Matriz de Confusão para Métrica

axs[1,0].scatter(X_train_MinMax[:,0],X_train_MinMax[:,1],c=Y_train_MinMax)
axs[1,0].set_title('Training MinMax Preprocessing')
axs[1,0].grid()

axs[1,1].scatter(X_test_MinMax[:,0],X_test_MinMax[:,1],c=Y_test_MinMax)
axs[1,1].set_title('Test MinMax Preprocessing')
axs[1,1].grid()

axs[1,2].scatter(X_test_MinMax[:,0],X_test_MinMax[:,1],c=Y_pred_MinMax)
axs[1,2].set_title('Predict MinMax Preprocessing')
axs[1,2].grid()

# Normalização L1 Matriz de Confusão para Métrica

axs[2,0].scatter(X_train_L1[:,0],X_train_L1[:,1],c=Y_train_L1)
axs[2,0].set_title('Training L1 Preprocessing')
axs[2,0].grid()

axs[2,1].scatter(X_test_L1[:,0],X_test_L1[:,1],c=Y_test_L1)
axs[2,1].set_title('Test L1 Preprocessing')
axs[2,1].grid()

axs[2,2].scatter(X_test_L1[:,0],X_test_L1[:,1],c=Y_pred_L1)
axs[2,2].set_title('Predict L1 Preprocessing')
axs[2,2].grid()

# Normalização L2 Matriz de Confusão para Métrica

axs[3,0].scatter(X_train_L2[:,0],X_train_L2[:,1],c=Y_train_L2)
axs[3,0].set_title('Training L2 Preprocessing')
axs[3,0].grid()

axs[3,1].scatter(X_test_L2[:,0],X_test_L2[:,1],c=Y_test_L2)
axs[3,1].set_title('Test L2 Preprocessing')
axs[3,1].grid()

axs[3,2].scatter(X_test_L2[:,0],X_test_L2[:,1],c=Y_pred_L2)
axs[3,2].set_title('Predict L2 Preprocessing')
axs[3,2].grid()

plt.show()
````

````python
# Análise dos resultados: Métricas de Acurácia e Matriz de Confusão

from sklearn.metrics import accuracy_score

result = accuracy_score(Y_test,Y_pred)
print("Accuracy Standard:",result)

result_normal = accuracy_score(Y_test_normal,Y_pred_normal)
print("Accuracy Normal:",result_normal)

result_MinMax = accuracy_score(Y_test_MinMax,Y_pred_MinMax)
print("Accuracy MinMax:",result_MinMax)

result_L1 = accuracy_score(Y_test_L1,Y_pred_L1)
print("Accuracy L1:",result_L1)

result_L2 = accuracy_score(Y_test_L2,Y_pred_L2)
print("Accuracy L2:",result_L2)
````
